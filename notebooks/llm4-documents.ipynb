# LLM 4 — Give the AI your documents

The model doesn't know your club's rules, your syllabus, or your notes. Today: put YOUR text in front of it — the trick behind every 'chat with your docs' product.

> **Working with your AI pair:** paste any error into your AI and ask *"explain this error like I'm new, then help me fix it — don't just give me the answer."*


```
%pip install -q anthropic
import os, anthropic
from getpass import getpass
os.environ.setdefault('ANTHROPIC_API_KEY', getpass('Class API key: '))
MODEL='claude-opus-5'
client=anthropic.Anthropic()
def ask(prompt, max_tokens=500):
    r=client.messages.create(model=MODEL,max_tokens=max_tokens,
        messages=[{'role':'user','content':prompt}])
    return ''.join(b.text for b in r.content if b.type=='text')
```

## 1. A document the model has never seen


```
handbook = """ROBOTICS CLUB HANDBOOK (excerpt)
Meetings: Tuesdays and Thursdays, 4:00-5:30, room 214.
Attendance: miss three in a row without notice and you lose your build-team spot.
Tools: soldering only with Mr. Ortiz present. Safety glasses always.
Competitions: regional in March at UIC; travel forms due two weeks before.
Dues: $20 per semester; waivers available, just ask, no questions."""
with open("handbook.txt","w") as f: f.write(handbook)
print("saved")
```

## 2. Context stuffing — the honest name for it


```
doc = open("handbook.txt").read()
q = "I missed two meetings, am I off the team? And when is the competition?"
print(ask(f"""Answer using ONLY the document below. If the answer is not in it, say so.

DOCUMENT:
{doc}

QUESTION: {q}"""))
```

## 3. The two rules that make this trustworthy
**Only-from-the-document** stops hallucinated policy. **Say-when-it's-missing** stops invented answers. Test rule 2:


```
print(ask(f"Answer ONLY from this document, and say if it is not in there.\n\n{doc}\n\nQUESTION: Can freshmen join?"))
```

## 4. When documents get big
The context window (course 1!) has limits, and big prompts cost more. Real products search for the relevant CHUNK first, then stuff only that. That's called retrieval — you just learned the concept; the plumbing comes later in your journey.
