## 1. The dataset


```
import pandas as pd, random
random.seed(7)
hoods=['Pilsen','Bronzeville','Rogers Park','Uptown','Hyde Park']
rows=[{'student':f'S{i:02d}','neighborhood':random.choice(hoods),
       'study_h':round(random.uniform(0,10),1),'score':0} for i in range(80)]
for r in rows: r['score']=round(min(100, 55+r['study_h']*4+random.uniform(-12,12)))
df=pd.DataFrame(rows)
df.head()
```

## 2. groupby — the question machine
*Average score by neighborhood* is one line:


```
df.groupby('neighborhood')['score'].mean().round(1)
```

## 3. More shapes of question


```
print(df['score'].mean().round(1), 'overall average')
print(df.groupby('neighborhood')['study_h'].mean().round(1))
print(df[df['study_h']>5]['score'].mean().round(1),'avg score, heavy studiers')
print(df[df['study_h']<2]['score'].mean().round(1),'avg score, light studiers')
```

## 4. Correlation is not proof
Heavy studiers score higher here. Does studying CAUSE the score? In this fake data, yes — we built it that way. In real data you can't see the cause in the table. Say 'associated with', not 'causes', unless you ran an experiment.
