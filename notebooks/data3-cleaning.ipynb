## 1. A deliberately messy dataset


```
csv = """name,neighborhood,hours,signup
Maya,pilsen,3.5,2026-01-04
DeShawn,Bronzeville,,2026-01-05
Alex,ROGERS PARK,4.1,2026-01-05
Maya,pilsen,3.5,2026-01-04
Jordan,Little village,2.0,not sure
Sam,Uptown,4.9,2026-01-08
Riley,uptown,-1,2026-01-09"""
with open("messy.csv","w") as f: f.write(csv)
import pandas as pd
df = pd.read_csv("messy.csv")
df
```

## 2. Find the problems before fixing anything
Professionals *diagnose first*.


```
print(df.isna().sum())        # missing values per column
print()
print(df['neighborhood'].unique())   # spelling chaos
print()
print(df.duplicated().sum(),'duplicate rows')
```

## 3. Fix, one problem at a time — and say what you did


```
df = df.drop_duplicates()
df['neighborhood'] = df['neighborhood'].str.title().str.strip()
df.loc[df['hours']<0,'hours'] = None   # -1 hours is impossible → treat as missing
df
```

## 4. The judgment call
What do we DO with missing hours? Drop the row? Fill with the average? **There is no automatic right answer — you decide and you disclose.**


```
filled = df.fillna({'hours':df['hours'].mean()})
filled
```
