# Lesson 3 — Lists and loops

A **list** holds many values in order. A **loop** does something to each of them. Together they're half of programming.

## 1. A list


```
crew = ["Maya", "DeShawn", "Alex", "Jordan"]
print(crew)
print(crew[0])   # counting starts at ZERO
print(len(crew))
```

## 2. The for loop — do it for each one


```
crew = ["Maya", "DeShawn", "Alex", "Jordan"]
for person in crew:
    print(f"{person} is on the team")
```

**The indent matters.** The indented line is what repeats. Try adding a second indented line.


## 3. range — loop a number of times


```
for i in range(5):
    print(f"lap {i}")
```

Remember lesson 1's build? `range(100)` is how you print something 100 times. Now you know the machinery.


## 4. Loops that count and collect


```
total = 0
for n in [12, 7, 30, 1]:
    total = total + n
print(total)
```
