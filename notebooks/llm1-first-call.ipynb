## 3. Teacher-set model
One model for the whole class, so results are comparable. Every call costs real money — small `max_tokens` keeps it cheap.


```
MODEL = 'claude-opus-5'   # your teacher may give you a different model id
```

## 4. The first call


```
import anthropic
client = anthropic.Anthropic()

response = client.messages.create(
    model=MODEL,
    max_tokens=300,
    messages=[{'role':'user','content':'Explain in two sentences why Chicago is called the Windy City.'}]
)
for block in response.content:
    if block.type == 'text':
        print(block.text)
```

Notice the shape: a **list of messages** goes in; a **list of content blocks** comes back — always check the block's `.type`. And check `response.stop_reason`: `'end_turn'` means it finished; `'max_tokens'` means you cut it off; `'refusal'` means the manners school said no (course 1!).


```
print(response.stop_reason)
print(response.usage.input_tokens, 'tokens in /', response.usage.output_tokens, 'tokens out')
```
