## 1. Write a file


```
with open("note.txt", "w") as f:
    f.write("This file was created by my code.\n")
    f.write("Line two.\n")
print("saved!")
```

Click the 📁 folder icon on the left — `note.txt` is there. Your code made a real file.


## 2. Read it back


```
with open("note.txt", "r") as f:
    content = f.read()
print(content)
```

## 3. Append — add without erasing


```
with open("note.txt", "a") as f:
    f.write("A third line, added later.\n")

with open("note.txt") as f:
    print(f.read())
```

## 4. Loop over lines


```
with open("note.txt") as f:
    for i, line in enumerate(f, 1):
        print(f"{i}: {line.strip()}")
```
