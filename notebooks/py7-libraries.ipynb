## 1. random — chance


```
import random

print(random.randint(1, 6))      # a die
print(random.choice(["heads", "tails"]))
```

Run it several times. Different every time — that's the point.


## 2. datetime — time


```
from datetime import date

today = date.today()
print(today)

bday = date(2026, 12, 25)
print(f"Days until: {(bday - today).days}")
```

## 3. collections — a better tally
Remember building the tally by hand in lesson 5? A library does it in one line. (Both are worth knowing — you understand what the shortcut does.)


```
from collections import Counter
votes = ["pizza", "tacos", "pizza", "sushi", "pizza"]
print(Counter(votes))
print(Counter(votes).most_common(1))
```
