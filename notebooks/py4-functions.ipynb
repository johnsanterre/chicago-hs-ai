A **function** is code with a name. Write it once, use it forever. Every tool you've called — `print()`, `len()`, `range()` — is a function someone wrote for you. Today you write your own.

## 1. Define, then call


```
def cheer(name):
    print(f"Go {name}, go!")

cheer("Maya")
cheer("DeShawn")
```

## 2. return — functions that hand back a value


```
def area(width, height):
    return width * height

room = area(12, 10)
print(f"The room is {room} square feet")
```

## 3. Functions calling functions


```
def area(width, height):
    return width * height

def cost(width, height, price_per_sqft):
    return area(width, height) * price_per_sqft

print(cost(12, 10, 4.50))
```
