# Lesson 5 — Dictionaries

A **dictionary** stores pairs: a key and its value. It's how programs remember *facts about things*.

## 1. Key → value


```
player = {"name": "Maya", "team": "Wildcats", "points": 31}
print(player["name"])
print(player["points"])
```

## 2. Add, change


```
player = {"name": "Maya", "points": 31}
player["points"] = player["points"] + 12
player["mvp"] = True
print(player)
```

## 3. Loop over a dictionary


```
scores = {"Maya": 31, "DeShawn": 28, "Alex": 40}
for name in scores:
    print(f"{name} scored {scores[name]}")
```

## 4. The tally pattern — count things
This pattern appears everywhere: counting words, votes, anything.


```
votes = ["pizza", "tacos", "pizza", "sushi", "pizza", "tacos"]
tally = {}
for v in votes:
    tally[v] = tally.get(v, 0) + 1
print(tally)
```
