```
%pip install -q anthropic
import os, anthropic
from getpass import getpass
os.environ.setdefault('ANTHROPIC_API_KEY', getpass('Class API key: '))
MODEL='claude-opus-5'
client=anthropic.Anthropic()
```

## 1. Why the history list is everything
The API is stateless — it remembers NOTHING between calls (course 1: the whole conversation is re-read every turn). Your list IS the memory.


```
history = []
system = 'You are a friendly study buddy for a Chicago high-schooler. Short answers. Ask a follow-up question when it helps.'

def chat_turn(user_text):
    history.append({'role':'user','content':user_text})
    r = client.messages.create(model=MODEL, max_tokens=400,
                               system=system, messages=history)
    reply = ''.join(b.text for b in r.content if b.type=='text')
    history.append({'role':'assistant','content':reply})
    return reply
```

## 2. Prove the memory works


```
print(chat_turn('My name is Maya and my exam is on photosynthesis.'))
print()
print(chat_turn('Quiz me on my exam topic — and use my name.'))
```

It used the name and the topic from turn one — because YOUR list carried them. Comment out the two `history.append` lines and try again to watch the amnesia.


## 3. The live loop


```
# Run this cell for a real conversation; type 'quit' to stop.
while True:
    msg = input('you: ')
    if msg.lower() in ('quit','q'): break
    print('bot:', chat_turn(msg), '\n')
```
