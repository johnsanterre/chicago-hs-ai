## 1. Data


```
import pandas as pd, matplotlib.pyplot as plt
counts = pd.Series({'Pilsen':21,'Bronzeville':17,'Rogers Park':14,'Uptown':9,'Hyde Park':19})
```

## 2. A bar chart, done honestly
Every honest chart has: a title that states the point, labeled axes, and a y-axis that starts at zero for bars.


```
counts.sort_values().plot(kind='barh')
plt.title('Sign-ups by neighborhood — Pilsen leads')
plt.xlabel('students signed up')
plt.tight_layout()
plt.show()
```

## 3. A line chart for change over time


```
weeks=[1,2,3,4,5,6]; signups=[4,9,15,28,41,80]
plt.plot(weeks,signups,marker='o')
plt.title('Sign-ups doubled almost every week')
plt.xlabel('week'); plt.ylabel('total sign-ups')
plt.ylim(0,None)
plt.show()
```
