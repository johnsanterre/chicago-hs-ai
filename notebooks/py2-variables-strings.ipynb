A **variable** is a name with a value stored in it. Once a value has a name, you can use it everywhere.

## 1. Names for numbers


```
age = 15
year = 2026
born = year - age
print(born)
```

Change `age` to your real age and run again. Notice: change one line, everything downstream updates.


## 2. Names for text
Text in quotes is called a **string**.


```
name = "Ada"
school = "Lane Tech"
print(name + " goes to " + school)
```

## 3. f-strings — the good way to mix text and values
Put an `f` before the quotes and drop variables into `{curly braces}`.


```
name = "Ada"
age = 15
print(f"{name} is {age} years old and will be 100 in {100 - age} years.")
```

## 4. Strings have superpowers
Run each line. Guess what it does before you run it.


```
word = "Chicago"
print(word.upper())
print(word.lower())
print(len(word))
print(word * 3)
```
