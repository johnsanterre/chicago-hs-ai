Curate a corpus, train your machine on it, tune it, generate a page, and

**The corpus rule: your words, or words old enough to belong to everyone.**
Your own essays, journals, and stories are perfect. Public-domain books
(gutenberg.org) are fine. Your group chat is NOT only your words, and
copyrighted books are not yours to train on.

```python
# ==================== PASTE YOUR CORPUS HERE ====================
my_corpus_raw = """
(Replace this with your text - the more the better. A few pages minimum;
old essays, stories, journal entries. It never leaves this notebook.)
"""

import re
corpus = re.sub(r"[^a-z ]+", " ", my_corpus_raw.lower())
corpus = re.sub(r" +", " ", corpus).strip()
print(len(corpus), "characters after cleaning")
if len(corpus) < 2000:
    print("NOTE: under ~2,000 characters the model will be very repetitive. "
          "More text = more voice.")
```

```python
# The machine, assembled from lessons 2-4: n-gram counting + temperature.
import random

def build(c, k):
    table = {}
    for i in range(len(c) - k):
        ctx, nxt = c[i:i+k], c[i+k]
        table.setdefault(ctx, {}).setdefault(nxt, 0)
        table[ctx][nxt] += 1
    return table

def generate(table, k, T=1.0, length=400):
    cur = corpus[:k]
    out = cur
    for _ in range(length):
        row = table.get(cur)
        if not row:
            out += " "
            cur = out[-k:]
            continue
        letters = list(row.keys())
        weights = [n ** (1.0 / T) for n in row.values()]
        nxt = random.choices(letters, weights=weights)[0]
        out += nxt
        cur = out[-k:]
    return out
```

```python
# Tune: try context lengths and temperatures until the voice sounds most
# like yours. (Longer context needs more corpus - lesson 4's wall applies.)
for k in [2, 3, 4]:
    table = build(corpus, k)
    print(f"=== context {k}, temperature 0.8")
    print(generate(table, k, T=0.8, length=200))
    print()
```

```python
# Your page: lock in your chosen settings and generate the real thing.
K = 3        # your choice
T = 0.8      # your choice
table = build(corpus, K)
page = generate(table, K, T=T, length=1200)
print(page)
```
