# LLM 2 — Prompts inside programs

A prompt in a program is an f-string with a job. The four levers from course 1 — context, task, format, limits — now live in code.

> **Working with your AI pair:** paste any error into your AI and ask *"explain this error like I'm new, then help me fix it — don't just give me the answer."*


```
%pip install -q anthropic
import os, anthropic
from getpass import getpass
os.environ.setdefault('ANTHROPIC_API_KEY', getpass('Class API key: '))
MODEL='claude-opus-5'
client = anthropic.Anthropic()
def ask(prompt, system=None, max_tokens=400):
    kwargs = dict(model=MODEL, max_tokens=max_tokens,
                  messages=[{'role':'user','content':prompt}])
    if system: kwargs['system'] = system
    r = client.messages.create(**kwargs)
    return ''.join(b.text for b in r.content if b.type=='text')
```

## 1. The system prompt — the standing instruction
The system prompt sets who the AI is for EVERY message. Users can't see it. This is how apps make one model behave a thousand ways.


```
coach = 'You are a writing coach for high-school students. Never rewrite their work. Point at weaknesses and ask questions that lead them to fixes. Be encouraging but honest.'
print(ask('Here is my opening sentence: "In this essay I will talk about pollution."', system=coach))
```

## 2. Prompts from variables — the template pattern


```
def flashcard(topic, level):
    return ask(f'Write one quiz question about {topic} for a {level} student, then the answer on a new line labeled ANSWER:. Nothing else.')
print(flashcard('the water cycle','9th grade'))
print(flashcard('tokens in language models','beginner'))
```

## 3. Batch — the loop meets the model


```
words = ['photosynthesis','tokenizer','due process']
for w in words:
    print(w.upper()); print(ask(f'Define {w} in one sentence a 15-year-old gets.', max_tokens=100)); print()
```
