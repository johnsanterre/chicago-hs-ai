Same numbers, opposite feelings. Today you build the lie so you can never be fooled by one.

## 1. The numbers


```
import matplotlib.pyplot as plt
months=['Jan','Feb','Mar','Apr']
score=[71,72,73,74]
```

## 2. The honest chart


```
plt.bar(months,score)
plt.ylim(0,100)
plt.title('Scores basically flat — up 3 points')
plt.show()
```

## 3. The same data, lying
One change: the y-axis now starts at 70.


```
plt.bar(months,score)
plt.ylim(70,75)
plt.title('SCORES SKYROCKET!!')
plt.show()
```

**That's the whole trick.** Truncated axis + excited title = a lie built from true numbers.


## 4. Lie #2 — the cherry-picked window


```
all_months=list(range(1,13))
attendance=[90,88,85,80,78,75,74,73,75,78,83,88]
plt.plot(all_months[6:9],attendance[6:9],marker='o')
plt.title('Attendance rising!')
plt.show()
plt.plot(all_months,attendance,marker='o')
plt.title('...inside a year-long decline and recovery')
plt.ylim(0,100)
plt.show()
```
