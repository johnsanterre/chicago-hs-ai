## 1. Make a tiny dataset
This cell writes a real CSV file — the universal table format.


```
csv_text = """name,neighborhood,age,favorite_food
Maya,Pilsen,15,tacos
DeShawn,Bronzeville,16,pizza
Alex,Rogers Park,15,pho
Jordan,Little Village,17,tamales
Sam,Uptown,16,jibarito"""
with open("crew.csv","w") as f:
    f.write(csv_text)
print("saved crew.csv")
```

## 2. Load it with pandas


```
import pandas as pd
df = pd.read_csv('crew.csv')
df
```

## 3. First questions every analyst asks


```
print(df.shape)      # (rows, columns)
print(df.columns.tolist())
df.head(3)
```

## 4. One column, one row


```
print(df['age'])
print()
print(df.loc[2])   # row number 2 — counting from zero, like lists
```
