# Lesson 8 — Mini-project: a program that does a chore

Everything you've learned, aimed at one target: build a small program that does a real chore for you or your family.

**Pick one** (or invent your own at this size):

- **The chore wheel** — fairly assigns this week's chores to the people in your house (lists, random, dictionaries)
- **The flashcard quizzer** — asks you questions from a file, keeps score (files, dictionaries, input)
- **The password maker** — generates strong passwords with rules you choose (random, strings, functions)
- **The allowance tracker** — logs money in/out to a file, prints a summary (files, f-strings, functions)

> **Working with your AI pair:** paste any error message into your AI assistant and ask *"explain this error like I'm new, then help me fix it — don't just give me the answer."*


## The rules of the build
1. **Design first:** in the cell below, write 4–6 comment lines describing what your program will do, step by step, *before any code*.
2. **Build in pieces:** one function at a time. Test each piece before the next.
3. **The AI is the intern:** it may write code for you, but you must be able to explain every line. If you can't explain it, make the AI explain it, then simplify it.
4. **It must actually run** top to bottom, twice in a row, without errors.
