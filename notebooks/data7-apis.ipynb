## 1. The city publishes its data
Chicago runs a public data portal: **data.cityofchicago.org** — crime, potholes, libraries, licenses. An **API** is a URL that returns data instead of a web page.


```
import requests, pandas as pd
url = 'https://data.cityofchicago.org/resource/x8fc-8rcq.json'  # Chicago public libraries
r = requests.get(url, timeout=30)
print(r.status_code)   # 200 means OK
data = r.json()
print(len(data),'records')
```

## 2. JSON → table
APIs speak JSON — dictionaries inside lists, exactly the nested shapes from the Python course.


```
df = pd.json_normalize(data)
print(df.columns.tolist()[:8])
df.head(3)
```
