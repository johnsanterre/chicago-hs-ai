## 1. A bigger table, generated by code


```
import pandas as pd, random
random.seed(3)
rows=[]
hoods=['Pilsen','Bronzeville','Rogers Park','Uptown','Little Village','Hyde Park']
for i in range(60):
    rows.append({'student':f'S{i:02d}','neighborhood':random.choice(hoods),
                 'hours_online':round(random.uniform(0.5,6),1),'grade':random.choice(list('ABBCCCD'))})
df=pd.DataFrame(rows)
df.head()
```

## 2. Filtering — the question mark of data
`df[df['col'] condition]` keeps only matching rows.


```
pilsen = df[df['neighborhood']=='Pilsen']
print(len(pilsen),'students from Pilsen')
heavy = df[df['hours_online']>4]
print(len(heavy),'students online more than 4 hours')
```

## 3. Sorting


```
df.sort_values('hours_online',ascending=False).head(5)
```

## 4. Counting categories
`value_counts()` is the tally pattern from the Python course, done in one line.


```
df['neighborhood'].value_counts()
```
